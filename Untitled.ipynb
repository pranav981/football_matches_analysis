```python
import pandas as pd

# 1. Load CSV
df = pd.read_csv('SP1.csv')
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

# 2. Check missing values
print("Missing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0] if df.isnull().sum().sum() else "No missing values")

# 3. Check duplicates
dupes = df.duplicated().sum()
print(f"\nDuplicate rows: {dupes}")
df = df.drop_duplicates()

# 4. Create team-level statistics
# Reshape to one row per team per match (home + away combined)
home = df.rename(columns={
    'HomeTeam': 'Team', 'AwayTeam': 'Opponent', 'FTHG': 'GF', 'FTAG': 'GA',
    'HS': 'Shots', 'AS': 'ShotsAgainst', 'HST': 'SOT', 'AST': 'SOTAgainst',
    'HF': 'Fouls', 'AF': 'FoulsAgainst', 'HC': 'Corners', 'AC': 'CornersAgainst',
    'HY': 'Yellow', 'HR': 'Red'
})
home['Venue'] = 'Home'
home['Result'] = df['FTR'].map({'H': 'Win', 'A': 'Loss', 'D': 'Draw'})

away = df.rename(columns={
    'AwayTeam': 'Team', 'HomeTeam': 'Opponent', 'FTAG': 'GF', 'FTHG': 'GA',
    'AS': 'Shots', 'HS': 'ShotsAgainst', 'AST': 'SOT', 'HST': 'SOTAgainst',
    'AF': 'Fouls', 'HF': 'FoulsAgainst', 'AC': 'Corners', 'HC': 'CornersAgainst',
    'AY': 'Yellow', 'AR': 'Red'
})
away['Venue'] = 'Away'
away['Result'] = df['FTR'].map({'H': 'Loss', 'A': 'Win', 'D': 'Draw'})

keep = ['Date', 'Team', 'Opponent', 'Venue', 'GF', 'GA', 'Shots', 'ShotsAgainst',
        'SOT', 'SOTAgainst', 'Fouls', 'FoulsAgainst', 'Corners', 'CornersAgainst',
        'Yellow', 'Red', 'Result']
team_stats = pd.concat([home[keep], away[keep]], ignore_index=True)

# 5. Create points
team_stats['Points'] = team_stats['Result'].map({'Win': 3, 'Draw': 1, 'Loss': 0})

# 6. Create wins/draws/losses summary per team
summary = team_stats.groupby('Team').agg(
    MatchesPlayed=('Result', 'count'),
    Wins=('Result', lambda x: (x == 'Win').sum()),
    Draws=('Result', lambda x: (x == 'Draw').sum()),
    Losses=('Result', lambda x: (x == 'Loss').sum()),
    Points=('Points', 'sum'),
    GoalsFor=('GF', 'sum'),
    GoalsAgainst=('GA', 'sum'),
).reset_index()
summary['GoalDiff'] = summary['GoalsFor'] - summary['GoalsAgainst']
summary['WinPct'] = round(summary['Wins'] / summary['MatchesPlayed'] * 100, 1)
summary = summary.sort_values('Points', ascending=False).reset_index(drop=True)

print("\nTeam standings (top 5):")
print(summary.head())

# Save outputs for Power BI
team_stats.to_csv('team_stats_long.csv', index=False)
summary.to_csv('team_summary.csv', index=False)
print("\nSaved: team_stats_long.csv, team_summary.csv")
```

```python
# =========================================================
# 2. TEAM FOUL ANALYSIS
# =========================================================

team_fouls = (
    team_stats.groupby('Team')
    .agg(
        Matches=('Fouls', 'count'),
        TotalFouls=('Fouls', 'sum'),
        FoulsPerMatch=('Fouls', 'mean'),
        Yellows=('Yellow', 'sum'),
        Reds=('Red', 'sum')
    )
    .reset_index()
)

team_fouls['FoulsPerMatch'] = (
    team_fouls['FoulsPerMatch'].round(2)
)

team_fouls['CardsPerMatch'] = (
    (team_fouls['Yellows'] + team_fouls['Reds'])
    / team_fouls['Matches']
).round(2)


# =========================================================
# 3. FOUL RANKING
# =========================================================

# Rank 1 = most fouls per match
team_fouls['FoulRank'] = (
    team_fouls['FoulsPerMatch']
    .rank(
        ascending=False,
        method='min'
    )
    .astype(int)
)

team_fouls = (
    team_fouls
    .sort_values('FoulRank')
    .reset_index(drop=True)
)


# =========================================================
# 4. FOUL CATEGORY
# =========================================================

n = len(team_fouls)

team_fouls['Category'] = 'Average'

team_fouls.loc[
    team_fouls['FoulRank'] <= 3,
    'Category'
] = 'Most Fouling'

team_fouls.loc[
    team_fouls['FoulRank'] > n - 3,
    'Category'
] = 'Most Disciplined'


# =========================================================
# 5. ATTACK & DEFENCE RANKINGS
# =========================================================

rankings = summary[
    [
        'Team',
        'MatchesPlayed',
        'GoalsFor',
        'GoalsAgainst',
        'Points'
    ]
].copy()

rankings['GoalsPerMatch'] = (
    rankings['GoalsFor']
    / rankings['MatchesPlayed']
).round(2)

rankings['ConcededPerMatch'] = (
    rankings['GoalsAgainst']
    / rankings['MatchesPlayed']
).round(2)


# Attack: more goals = better rank
rankings['AttackRank'] = (
    rankings['GoalsFor']
    .rank(
        ascending=False,
        method='min'
    )
    .astype(int)
)


# Defence: fewer goals conceded = better rank
rankings['DefenceRank'] = (
    rankings['GoalsAgainst']
    .rank(
        ascending=True,
        method='min'
    )
    .astype(int)
)


# =========================================================
# 6. MERGE FOUL DATA WITH TEAM RANKINGS
# =========================================================

team_rankings = rankings.merge(
    team_fouls[
        [
            'Team',
            'FoulsPerMatch',
            'Yellows',
            'Reds',
            'CardsPerMatch',
            'FoulRank',
            'Category'
        ]
    ],
    on='Team',
    how='left'
)


# =========================================================
# 7. DISPLAY RESULTS
# =========================================================

print("\nTeam Rankings:")
print(
    team_rankings
    .sort_values('AttackRank')
    .head(10)
)


# =========================================================
# 8. TOP PERFORMERS
# =========================================================

best_attack = (
    team_rankings[
        team_rankings['AttackRank'] == 1
    ].iloc[0]
)

best_defence = (
    team_rankings[
        team_rankings['DefenceRank'] == 1
    ].iloc[0]
)

most_fouling = team_fouls.iloc[0]
most_disciplined = team_fouls.iloc[-1]


print(
    f"\nBest attack      : "
    f"{best_attack['Team']} "
    f"({best_attack['GoalsFor']} goals)"
)

print(
    f"Best defence     : "
    f"{best_defence['Team']} "
    f"({best_defence['GoalsAgainst']} conceded)"
)

print(
    f"Most fouling     : "
    f"{most_fouling['Team']} "
    f"({most_fouling['FoulsPerMatch']} fouls/match)"
)

print(
    f"Most disciplined : "
    f"{most_disciplined['Team']} "
    f"({most_disciplined['FoulsPerMatch']} fouls/match)"
)


# =========================================================
# 9. HEAD-TO-HEAD ANALYSIS
# =========================================================

def head_to_head(team_a, team_b):

    h = team_stats[
        (team_stats['Team'] == team_a) &
        (team_stats['Opponent'] == team_b)
    ]

    h = h[
        [
            'Date',
            'Venue',
            'GF',
            'GA',
            'Result'
        ]
    ].sort_values('Date')

    print(
        f"\n{team_a} vs {team_b}: "
        f"W{(h['Result'] == 'Win').sum()} "
        f"D{(h['Result'] == 'Draw').sum()} "
        f"L{(h['Result'] == 'Loss').sum()}, "
        f"goals {h['GF'].sum()}-{h['GA'].sum()}"
    )

    return h


# Example
head_to_head(
    'Barcelona',
    'Real Madrid'
)


# =========================================================
# 10. SAVE NEW TABLES FOR POWER BI
# =========================================================

team_fouls.to_csv(
    'team_fouls.csv',
    index=False
)

team_rankings.to_csv(
    'team_rankings.csv',
    index=False
)

print("\nSaved successfully:")
print("team_fouls.csv")
print("team_rankings.csv")
```
